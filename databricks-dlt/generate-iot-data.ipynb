```python
import json
from datetime import datetime, timedelta
import random

# Config
num_devices = 10
total_records = 5000
start_time = datetime(2025, 1, 1, 0, 0, 0)
devices = [f"device_{i+1}" for i in range(num_devices)]

# Schema
schema = ["device_id", "timestamp", "temperature", "humidity", "status"]

# Generate data
batch_data = []
for _ in range(total_records):
    device_id = random.choice(devices)
    timestamp = start_time + timedelta(seconds=random.randint(0, 7*24*60*60))  # 1 week
    
    # Simulate realistic sensor data
    temp_base = 25 + random.uniform(-10, 10)  # Room temp ±10
    temperature = round(temp_base + random.uniform(-5, 5), 2)
    
    humidity = round(random.uniform(30, 80), 2)
    
    # 5% bad  null device_id, extreme temp
    if random.random() < 0.05:
        device_id = None
    if random.random() < 0.05:
        temperature = 999  # Bad reading
    
    status = random.choice(["active", "maintenance", "offline"])
    
    record = {
        "device_id": device_id,
        "timestamp": timestamp.isoformat(),
        "temperature": temperature,
        "humidity": humidity,
        "status": status
    }
    batch_data.append(record)

# Write to Volume
output_path = "/Volumes/raw/default/raw_data_volume/iot/simulated_iot_data.json"
with open(output_path, "w") as f:
    for row in batch_data:
        f.write(json.dumps(row) + "\n")

print(f"✅ Wrote {len(batch_data)} IoT records to {output_path}")
```
